```python
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import StandardScaler

```

#### **1. Loading and Preprocessing **
##### ● Load the California Housing dataset using the fetch_california_housing function from sklearn.
##### ● Convert the dataset into a pandas DataFrame for easier handling. Handle missing values (if any) and perform necessary feature scaling (e.g., standardization).  

```python
housing_data = fetch_california_housing()
df = pd.DataFrame(housing_data.data, columns=housing_data.feature_names)
```

```python
print(housing_data.data)
```

```python
print(df.columns)
```

```python
df.dtypes
```

```python
df['Price'] = housing_data.target
```

##### Check Missing values

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df.shape
```

```python
# Seperate feature and target
x = df.drop('Price', axis=1)
y = df['Price']
```

```python
# Feature Scaling
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
# fit calculates the mean and standard deviation of each feature
# transform() convert it into a scaled value.
# StandardScaler return a NumPy array, so return it back to dataframe
x_scaled = pd.DataFrame(x_scaled,columns = x.columns)
```

##### **Explain the preprocessing steps you performed and justify why they are necessary for this dataset.**

##### First loaded the dataset and converted it into DataFrame. Then missing values are checked and found that there was no missing values. Regression model need independent and target variables, so seperated targeted variable. Feature scaling was performed using StandardScaler because different features have different ranges.


#### **2. Regression Algorithm Implementation**
##### ● Implement the following regression algorithms:
#####    ○ Linear Regression  
#####    ○ Decision Tree Regressor  
##### ● For each algorithm: Provide a brief explanation of how it works. Explain why it might be suitable for this dataset.  

```python
# Linear Regression
# used for divide dataset into training and testing data
from sklearn.model_selection import train_test_split
# import linear regression algorithm
from sklearn.linear_model import LinearRegression
# Split dataset into training (80%) and testing data (20%)
x_train, x_test, y_train, y_test = train_test_split(x_scaled,y,test_size=0.2,random_state=42)
# random _state =42 ensures the same spllit every time.
lr_model = LinearRegression() # create linear regression model
lr_model.fit( x_train, y_train) # train model. Model learn the relationship between dependent and independent variales.
lr_predict = lr_model.predict( x_test) # Predict the house price
print(lr_predict)
```

#####  Linear regression is a simple algorithm. Here in this dataset, it identifies the relationship between the housing features and price. By providing 80% of training it learned the model well and predict the house price.

```python
# Decision Tree Regressor
from sklearn.tree import DecisionTreeRegressor # Import DecisionTreeRenerator algorithm
dt_model = DecisionTreeRegressor(random_state=42) # Create DecisionTreeRenerator model
dt_model.fit(x_train,y_train) # It split data into decision rules.
dt_predict = dt_model.predict(x_test) # Predict the value
print(dt_predict)
```

##### Rather than linear relation decision trees can consider nonlinear relationships to predict the house price

#### Model Evaluation and Comparison
##### ● Evaluate the performance of each algorithm using the following metrics:
##### ○ Mean Squared Error (MSE)  
##### ○ Mean Absolute Error (MAE)  
##### ○ R-squared Score (R²)  

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
result = {}
```

```python
# Evaluate linear Regression Model
result['Linear Regression'] = [
    mean_squared_error(y_test, lr_predict),
    mean_absolute_error(y_test, lr_predict),
    r2_score(y_test, lr_predict)
]
```

```python
# Decision Tree Regressor
result['Decision Tree Regressor'] = [
    mean_squared_error(y_test, dt_predict),
    mean_absolute_error(y_test, dt_predict),
    r2_score(y_test, dt_predict)
]
```

```python
# Convert Results into DataFrame
results_df = pd.DataFrame(
    result,
    index=['MSE','MAE','R2 Score']
)
print(results_df)
```

##### **Compare the results of all models and identify: The best-performing algorithm with justification.**
##### The Decision Tree Regressor is the best-performing model among the tested algorithms. It achieved the lowest MSE and MAE values and the highest R² score, indicating better prediction accuracy. The model can capture non-linear relationships between housing features and house prices, which improves its performance compared to Linear Regression.
